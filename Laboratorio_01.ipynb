```python
import pandas as pd
```

```python
data = pd.read_csv('/content/SOCR-HeightWeight.csv')
data
```

```python
# Mostrar cantidad de registros y atributos
print(f"Cantidad de registros: {data.shape[0]}")
print(f"Cantidad de atributos: {data.shape[1]}")
```

```python
# b. Determinar cantidad de variables numéricas y no numéricas
num_vars = data.select_dtypes(include=['number']).shape[1]
cat_vars = data.select_dtypes(exclude=['number']).shape[1]
missing_values = data.isnull().sum().sum()

print(f"Variables numéricas: {num_vars}")
print(f"Variables no numéricas: {cat_vars}")
print(f"Valores perdidos en total: {missing_values}")
```

```python
# c. Calcular estadísticos descriptivos para variables numéricas
stats = data.describe()

print("\nEstadísticos Descriptivos:")
print(stats)
```
